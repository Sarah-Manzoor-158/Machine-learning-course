1.Numpy

```python
import numpy as np
float_array = np.array([1, 2, 3, 4], dtype=float)
print("Float Array:", float_array)
zeros_array = np.zeros((3, 4))
print("Zeros Array:", zeros_array)
my_tuple = (10, 20, 30, 40)
tuple_array = np.array(my_tuple)
print("Tuple Array:", tuple_array)
random_array = np.random.rand(3, 3)
print("Random Array:", random_array)
```

```python
import numpy as np
original_array = np.arange(12).reshape(3, 4)
reshaped_array = original_array.reshape(2, 2, 3)
print("Reshaped Array:", reshaped_array)
step_sequence = np.arange(0, 31, 5)
print("Step Sequence:", step_sequence)
flattened_array = reshaped_array.flatten()
print("Flattened Array:", flattened_array)
complex_array = np.full((2, 3), fill_value=3 + 4j, dtype=complex)
print("Complex Constant Array:", complex_array)
```
