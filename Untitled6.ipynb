```python
import pandas as pd
df = pd.read_csv('/content/Building_Permits.csv')
print(df.head())
```

```python
total_values = df.size
missing_values = df.isnull().sum().sum()
percentage = (missing_values / total_values) * 100
print("Total missing values:", missing_values)
print("Percentage of missing values:", percentage, "%")
```

```python
df_drop = df.dropna()
print(df_drop.head())
print("Shape before:", df.shape)
print("Shape after:", df_drop.shape)
```

```python
df_fill = df.copy()
for col in df_fill.columns:
    if df_fill[col].dtype == 'object':
        if df_fill[col].mode().empty:
            df_fill[col] = df_fill[col].fillna('Unknown')
        else:
            df_fill[col] = df_fill[col].fillna(df_fill[col].mode()[0])
    else:
        df_fill[col] = df_fill[col].fillna(df_fill[col].median())
print(df_fill.head())
print("Missing values after filling:")
print(df_fill.isnull().sum().sum())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.preprocessing import minmax_scaling
```

```python
df = pd.read_csv('/content/ks-projects-201801.csv')
print(df.head())
```

```python
usd_goal = df[['usd_goal_real']].dropna()
usd_goal_normalized = minmax_scaling(
    usd_goal,
    columns=['usd_goal_real']
)
print(usd_goal_normalized.head())
```

```python
goal = df[['goal']].dropna()
goal_normalized = minmax_scaling(
    goal,
    columns=['goal']
)
print(goal_normalized.head())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(usd_goal_normalized['usd_goal_real'], bins=30)
plt.title('Normalized USD Goal Real')
plt.xlabel('Normalized Values')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(goal_normalized['goal'], bins=30)
plt.title('Normalized Goal')
plt.xlabel('Normalized Values')
plt.ylabel('Frequency')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv('/content/Pakistan_Largest_Ecommerce_Dataset.csv')
print(df.head())
```

```python
data = df[['price', 'grand_total']].dropna()
print(data.head())
```

```python
scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(data)
normalized_df = pd.DataFrame(
    normalized_data,
    columns=['price', 'grand_total']
)
print(normalized_df.head())
```

```python
normalized_df.hist(figsize=(10, 5), bins=30)
plt.suptitle('Normalized Price and Grand Total')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('/content/earthquake_database.csv')
print(df.head())
```

```python
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%m/%d/%Y',
    errors='coerce'
)
print(df['Date'].head())
```

```python
df['Day'] = df['Date'].dt.day
print(df[['Date', 'Day']].head())
```

```python
df = df.dropna(subset=['Day'])
df['Day'] = df['Day'].astype(int)
print(df[['Date', 'Day']].head())
```

```python
sns.displot(
    data=df,
    x='Day',
    bins=31,
    kde=False
)
plt.title('Distribution of Earthquake Days')
plt.xlabel('Days of Month')
plt.ylabel('Frequency')
plt.show()
```
