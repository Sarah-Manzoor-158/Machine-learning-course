```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import io
import pandas as pd
from google.colab import files

uploaded = files.upload()

# Automatically assign DataFrames dynamically regardless of uploaded filename numbers
dfs = {}
for filename, content in uploaded.items():
    # Remove extension and clean up names like 'fifa (2).csv' -> 'fifa'
    clean_name = filename.split('.')[0].split(' ')[0]
    dfs[clean_name] = pd.read_csv(io.BytesIO(content))

# Now safely access your DataFrames:
fifa_df = dfs.get('fifa')
flight_delays_df = dfs.get('flight_delays')
insurance_df = dfs.get('insurance')
iris_df = dfs.get('iris')
spotify_df = dfs.get('spotify')

print("Successfully loaded datasets:", list(dfs.keys()))
```

```python
# Path of the file to read
fifa_filepath = "/content/fifa.csv"
# Read the file into a variable fifa_data
fifa_data = pd.read_csv(fifa_filepath, index_col="Date", parse_dates=True)
```

```python
fifa_data.head(6)
```

```python
# Set the width and height of the figure
plt.figure(figsize=(15,4))
# Line chart showing how FIFA rankings evolved over time
sns.lineplot(data=fifa_data)
```

```python
# Path of the file to read
spotify_filepath = "/content/spotify.csv"
# Read the file into a variable spotify_data
spotify_data = pd.read_csv(spotify_filepath, index_col="Date", parse_dates=True)
```

```python
# Path of the file to read
flight_filepath = "/content/flight_delays.csv"
# Read the file into a variable flight_data
flight_data = pd.read_csv(flight_filepath, index_col="Month")
```

```python
flight_data
```

```python
# Set the width and height of the figure
plt.figure(figsize=(7,14))
# Add title
plt.title("Average Arrival Delay for Spirit Airlines Flights, by Month")
# Bar chart showing average arrival delay for Spirit Airlines flights by month
sns.barplot(x=flight_data.index, y=flight_data['NK'])
# Add label for vertical axis
plt.ylabel("Arrival delay (in minutes)")
```

```python
# Set the width and height of the figure
plt.figure(figsize=(10,6))
# Add title
plt.title("Average Arrival Delay for Each Airline, by Month")
# Heatmap showing average arrival delay for each airline by month
sns.heatmap(data=flight_data, annot=True)
# Add label for horizontal axis
plt.xlabel("Airline")
```
