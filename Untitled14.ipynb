```python
import pandas as pd
ign_data = pd.read_csv("ign_scores.csv", index_col="Platform")
print(ign_data)
pc_max = ign_data.loc['PC'].max()
pc_genre = ign_data.loc['PC'].idxmax()
print(f"\nPC Highest Score: {pc_max} in {pc_genre}")
psv_min = ign_data.loc['PlayStation Vita'].min()
psv_genre = ign_data.loc['PlayStation Vita'].idxmin()
print(f"PS Vita Lowest Score: {psv_min} in {psv_genre}")
```

```python
plt.figure(figsize=(14, 6))
plt.title("Average IGN Score for Racing Games by Platform")
sns.barplot(x=ign_data.index, y=ign_data['Racing'])
plt.xticks(rotation=90)
plt.xlabel("Platform")
plt.ylabel("Average Rating")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
candy_data = pd.read_csv("candy.csv")
plt.figure(figsize=(10, 6))
sns.regplot(x=candy_data['sugarpercent'], y=candy_data['winpercent'])
plt.title("Sugar Percentage vs Win Percentage")
plt.show()
sns.lmplot(x="pricepercent", y="winpercent", hue="chocolate", data=candy_data, height=6, aspect=1.4)
plt.title("Price Percent vs Win Percent (Chocolate vs Non-Chocolate)")
plt.show()
plt.figure(figsize=(8, 6))
sns.swarmplot(x=candy_data['chocolate'], y=candy_data['winpercent'])
plt.title("Win Percent Distribution by Chocolate")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
cancer_data = pd.read_csv("cancer.csv")
plt.figure(figsize=(10, 6))
sns.histplot(data=cancer_data, x='Area (mean)', hue='Diagnosis')
plt.title("Histogram of Mean Area by Diagnosis")
plt.show()
plt.figure(figsize=(10, 6))
sns.kdeplot(data=cancer_data, x='Radius (worst)', hue='Diagnosis', shade=True)
plt.title("KDE Plot of Worst Radius by Diagnosis")
plt.show()
```
