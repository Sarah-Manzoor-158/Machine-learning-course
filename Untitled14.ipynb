```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fifa_data = pd.read_csv("fifa.csv", index_col="Date", parse_dates=True)
plt.figure(figsize=(14, 6))
plt.title("FIFA Rankings Over Time")
sns.lineplot(data=fifa_data)
plt.xlabel("Date")
plt.ylabel("Ranking")
plt.show()
```
